```python
max=5
queue=[]
def enqueue():
    if len(queue)==max:
        print("Queue overflow")
    else:
        item=int(input("enter a element"))
        queue.append(item)
        print("element inserted")
def dequeue():
    if len(queue)==0:
        print("Queue underflow")
    else:
        print("delect element:",queue.pop(0))
def display():
    if len(queue)==0:
        print("Queue is empty")
    else:
        print("Queue element:",queue)
while True:
    print("\n---QUEUE MENT---")
    print("1.enqueue")
    print("2. dequeue")
    print("3.display")
    print("4.EXIT")
    choice=int(input("enter a choice"))
    if choice==1:
        enqueue()
    elif choice==2:
        dequeue()
    elif choice==3:
        display()
    elif choice==3:
        break
    else:
        print("invalid choice")
```

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class Queue:
    def __init__(self):
        self.front=None
        self.rear=None
    def enqueue(self,data):
        new_node=Node(data)
        if self.rear is None:
            self.front=self.rear=new_node
        else:
            self.rear.next=new_node
            self.rear=new_node
        print("element insert successful")
    def dequeue(self):
        if self.front is None:
            print("queue underflow")
        else:
            data=self.front.data
            self.front=self.front.next
        if self.front is None:
            self.rear=None
        print("deleted element:",data)
    def display(self):
        if self.front is None:
            print("queue is empty")
        else:
            current=self.front
            print("queue element:",end='')
            while current:
                print(current.data,end='')
                current=current.next
            print()
queue=Queue()
while True:
    print("\n--queue ment--")
    print("1.enqueue")
    print("2.dequeue")    
    print("3.display")
    print("4.exit")
    choice=int(input("enter your choice:"))
    if choice==1:
        value=int(input("enter your element:"))
        queue.enqueue(value)
    elif choice==2:
        queue.dequeue()
    elif choice==3:
        queue.display()
    elif choice==4:
        break
    else:
        print("invalid")

        
    
    
    
    
    
        
        
           
        
```

```python
class Node:
    def __init__ (self,data):
        self.data=data
        self.left=None
        self.right=None
def inorder(root):
    if root:
        inorder(root.left)
        print(root.data,end='')
        inorder(root.right)
def preorder(root):
    if root:
        print(root.data,end='')
        preorder(root.left)
        preorder(root.right)
def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.data,end='')
root=Node(1)
root.left=Node(2)
root.right=Node(3)
root.left.left=Node(4)
root.left.right=Node(5)
print("inorder")
inorder(root)
print("\n preorder")
preorder(root)
print("\n postorder")
postorder(root)
            
               
```
